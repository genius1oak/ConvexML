## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

We would predict a **relevance score** for each individual flight option within a given search result set. This score represents the likelihood that a user would select or prefer that specific flight given their search criteria.

This label can come from two primary sources:

1.  **Observed Outcome**: The most direct signal would be **user booking data**. If a user books a particular flight, that flight receives a high relevance signal. Other observed behaviors like clicks, time spent viewing a flight, or adding to a shortlist could also serve as proxy labels, albeit with weaker signals.

2.  **Defined Rule/Proxy**: In cases where direct booking data is sparse or for initial model development, a relevance score could be a **composite proxy score** based on expert-defined rules. This score might combine factors like: price competitiveness, journey duration, number of layovers, airline reputation, preferred departure/arrival times, and consistency with typical user preferences for similar routes. This defined rule would act as a stand-in until sufficient observed outcome data is available for direct learning.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

For a flight ranking problem, **Normalized Discounted Cumulative Gain (NDCG)** is a robust and appropriate success metric. NDCG measures the quality of a ranking by considering the relevance of items at different positions. It assigns higher value to highly relevant items appearing at the top of the ranked list.

### What number means 'good'?

NDCG scores typically range from 0 to 1, where:

*   **1.0** represents a perfect ranking (all most relevant flights are at the top).
*   **0.0** represents the worst possible ranking (no relevant flights, or relevant flights are at the very bottom).

Ideally, we would aim for an **NDCG score as close to 1.0 as possible**, typically striving for values significantly above 0.8 to consider the ranking 'good' or effective in a production environment. The specific threshold for 'good' would depend on baseline performance, business requirements, and user experience targets.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Machine learning is essential for flight ranking because the patterns determining a user's flight preference are **highly complex, dynamic, and non-linear**, making them impossible to capture effectively with a simple fixed-rule system or if-statements.

Here's why:

1.  **Interactions and Trade-offs**: User decisions often involve intricate trade-offs between multiple features (e.g., a slightly higher price might be acceptable for a significantly shorter flight or a preferred airline). These interactions are hard to define explicitly with rules (e.g., 'if price is X, but duration is Y, and airline is Z, then rank higher'). ML models can learn these complex interactions automatically.
2.  **Dynamic Preferences and Market Conditions**: User preferences are not static. What's considered a 'good deal' or a 'convenient flight' can change based on the time of year, destination, origin, personal circumstances, and even current events. Flight prices, availability, and demand fluctuate constantly. A fixed rule system would require constant, manual updates and would quickly become obsolete.
3.  **Personalization**: Different users have different priorities. Some prioritize price, others convenience, others specific airlines or times. A fixed rule applies universally, but ML can learn and adapt to individual user preferences (given enough data), providing personalized rankings.
4.  **Implicit Signals**: User behavior (clicks, scroll depth, bookings) provides implicit signals about relevance that are difficult to encode in explicit rules. ML models can learn from these implicit feedback loops to refine their understanding of 'relevance'.
5.  **Feature Complexity and Scale**: The number of potential features (price, duration, stops, layover times, airline, aircraft type, time of day, day of week, lead time to booking, historical search patterns, etc.) and their possible values is enormous. Manually crafting rules that cover all these combinations and their nuanced effects would be an intractable task.
